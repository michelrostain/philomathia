## **Job 11**

Considérez un dé $D_6$ dont les valeurs possibles sont : $\{1, 2, 3, 4, 5, 6\}$.

Ce dé est non truqué, ce qui signifie que toutes les valeurs ont des probabilités égales de se produire.  

Si je jette le dé $D_6$, la probabilité que le résultat soit $3$ est de $\frac{1}{6}$.

1. Quelle est la probabilité d'obtenir $7$ au jet du dé $D_6$ ?
2. Quelle est la probabilité d'obtenir un résultat pair ?
3. Quelle est la probabilité d'obtenir un résultat impair ?
4. Quelle est la probabilité d'obtenir un résultat inférieur ou égal à $4$ ?
5. Quelle est la probabilité d'obtenir un résultat inférieur ou égal à $6$ ?
6. Quelle est la probabilité de ne pas obtenir $1$ ?

**1) Probabilité d'obtenir 7  au jet de dé à 6 faces :**       
Soit $X$ la variable aléatoire donnant le résultat d'un jet de $D_6$.
$X$ prend ses valeurs dans $\{1, \dots, 6\}$ et, pour tout $k \in \{1, \dots, 6\}$, $P(X = k) = \frac{1}{6}$.

Or $7 \notin \{1, \dots, 6\}$, donc l'événement $\{X = 7\}$ est impossible, c'est-à-dire $\{X = 7\} = \emptyset$.

Par conséquent :
$$P(X = 7) = 0$$

**2) et 3 ) Probabilité d'obtenir un nombre pair ou impair :**     
Soit $X$ la variable aléatoire donnant le résultat d'un jet de $D_6$.

L'événement « obtenir un résultat pair » est $A = \{X = 2\} \cup \{X = 4\} \cup \{X = 6\}$.

Ces trois événements sont deux à deux incompatibles (le dé ne peut afficher qu'une seule valeur), donc :

$$P(A) = P(X = 2) + P(X = 4) + P(X = 6) = \frac{1}{6} + \frac{1}{6} + \frac{1}{6} = \frac{3}{6} = \frac{1}{2}$$

```python
# Ici fraction sert à écrire un nombre rationnel en forme fraction.
from fractions import Fraction

# Nombres pairs

univers = {1,2,3,4,5,6}

proba = {k:Fraction(1,6) for k in univers}

A= {k for k in univers if k%2==0}

P_A = sum(proba[k] for k in A)

print(A)
print(P_A)


# Nombres impairs

univers = {1,2,3,4,5,6}

proba = {k:Fraction(1,6) for k in univers}

A= {k for k in univers if k%2==1}

P_A = sum(proba[k] for k in A)

print(A)
print(P_A)
```

**4) Obtention des éléments inférieurs ou égaux à 4**    
Soit $X$ la variable aléatoire donnant le résultat d'un jet de $D_6$.

L'événement « obtenir un résultat inférieur ou égal à 4 est $B = \{X =1 \} \cup \{X = 2\} \cup \{X = 3\} \cup \{X = 4\}$.

Ces quatre événements sont deux à deux incompatibles (le dé ne peut afficher qu'une seule valeur), donc :

$$P(A) = P(X = 1) + P(X = 2) + P(X = 3) + P(X=4) = \frac{1}{6} + \frac{1}{6} + \frac{1}{6} + \frac{1}{6}= \frac{4}{6} = \frac{2}{3}$$

**5) Obtention d'un résultat inférieur ou égal à 6 :**     
L'événement "obtenir un résultat inférieur ou égal à 6" est :      
$C = \{X=1 \} \cup \{X=2 \} \cup \{X=3 \} \cup \{X=4 \} \cup \{X=5 \} \cup \{X=6 \}$     

Ces six événements sont deux à deux incompatibles (le dé ne peut afficher qu'une seule valeur), donc :
$$P(C) = P(X = 1) + P(X = 2) + P(X = 3) + P(X=4) + P(X = 5) + P(X=6) = \frac{1}{6} + \frac{1}{6} + \frac{1}{6} + \frac{1}{6} + \frac{1}{6} + \frac{1}{6} = \frac{6}{6} = 1$$

C représente l'univers entier (les 6 possibilités) donc c'est un événement certain.    

Ecriture plus courte :      
$$C = \{X \le 6\} = \bigcup_{k=1}^{6} \{X=k\}$$
Evènements 2 à 2 incompatibles donc :
$$P(C) = \sum_{k=1}^{6} P(X=k) = 6 \times \frac{1}{6} = 1$$

$C$ est l'univers $\Omega$ : c'est un événement certain.

**Exemple : résultat inférieur ou égal à 4**

$$P(X \le 4) = \sum_{k=1}^{4} P(X=k) = 4 \times \frac{1}{6} = \frac{2}{3}$$
